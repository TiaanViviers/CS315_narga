# Visualise possible colors from the image, using a k-means clustering across the pixel space


```python
def plot_pixels(data, title, colors=None, N=10000):
    if colors is None:
        colors = data
    
    # choose a random subset
    rng = np.random.RandomState(0)
    i = rng.permutation(data.shape[0])[:N]
    colors = colors[i]
    R, G, B = data[i].T
    
    fig, ax = plt.subplots(1, 2, figsize=(16, 6))
    ax[0].scatter(R, G, color=colors, marker='.')
    ax[0].set(xlabel='Red', ylabel='Green', xlim=(0, 1), ylim=(0, 1))

    ax[1].scatter(R, B, color=colors, marker='.')
    ax[1].set(xlabel='Red', ylabel='Blue', xlim=(0, 1), ylim=(0, 1))

    fig.suptitle(title, size=20);
```

 The  MiniBatchKmeans is a variant of the KMeans algorithm which uses mini-batches to reduce the computation time, 
instead of using the full dataset at each iteration while still attempting to optimize the same objective function. 
 Mini-batches are subsets of the input data, randomly sampled in each training iteration. 
This increases the speed of the algorithm by a factor of 3–4 typically. Especially important, it makes it possible to cluster huge datasets that do not fit in memory.

Color Quantization is the process of reducing the number of colors in an image while keeping the visual appearance of the image intact. 
This is a useful image compression technique that is quite useful for devices that can show a limited number of colors due to memory restriction.

```python
from sklearn.cluster import MiniBatchKMeans
kmeans = MiniBatchKMeans(16)
kmeans.fit(data)
new_colors = kmeans.cluster_centers_[kmeans.predict(data)]

plot_pixels(data, colors=new_colors,
            title="Color space: 16 colors")
```

2. Assign class labels, i.e. assign each observation to the cluster mean closest to it
3.  Update cluster means, i.e. compute cluster means using the labeled (assigned) observations
4.  Repeat above two steps until convergence

## 3. K-means Decision Boundary
Suppose a K-means model using K = 2 for 2-dimensional data has means at A(10.7, 3.6) and B(1.7, 1.5). Assume the axis labels x1[horizontal] and x2[vertical].

Compute the equation for the decision boundary separating the regions corresponding to the two means, in the form 
$$x2 = mx1 + c$$
Provide the value of c.

```python
def kmeans_two_mean_boundary(mean_a, mean_b):
    """Return the linear decision boundary between two k-means means.

    Returns coefficients (w1, w2, c) for the line:
    w1*x1 + w2*x2 + c = 0
    """
    """Return m and c for the k-means boundary x2 = m*x1 + c."""
    mean_a = np.asarray(mean_a, dtype=float)
    mean_b = np.asarray(mean_b, dtype=float)

    w = mean_b - mean_a
    c0 = 0.5 * (np.dot(mean_a, mean_a) - np.dot(mean_b, mean_b))

    m = -w[0] / w[1]
    c = -c0 / w[1]

    return m, c
    
m, c = kmeans_two_mean_boundary([10.7, 3.6], [1.7, 1.5])
print(c)
```

## 6. Custom K-means Iterations
The following is a set of one-dimensional points: {1, 1, 2, 3, 5, 8, 13, 21, 33, 54}. 

Perform two complete iterations of K-means on these points using the two initial centroids u1 = 0 and u2 = 11. 

What is the value of u1 after having completed two iterations of the K-means algorithm.

```python
def kmeans_1d_iterations(points, initial_centroids, n_iter=2):
    """Run 1D k-means for a fixed number of full iterations."""
    points = np.asarray(points, dtype=float)
    centroids = np.asarray(initial_centroids, dtype=float).copy()

    history = []

    for _ in range(n_iter):
        distances = np.abs(points[:, np.newaxis] - centroids[np.newaxis, :])
        labels = np.argmin(distances, axis=1)

        new_centroids = centroids.copy()
        for cluster_index in range(len(centroids)):
            cluster_points = points[labels == cluster_index]
            if len(cluster_points) > 0:
                new_centroids[cluster_index] = np.mean(cluster_points)

        history.append({
            "labels": labels.copy(),
            "centroids": new_centroids.copy(),
        })
        centroids = new_centroids

    return history


points = [1, 1, 2, 3, 5, 8, 13, 21, 33, 54]
initial_centroids = [0, 11]
history = kmeans_1d_iterations(points, initial_centroids)
history[1]["centroids"]
```

## 1. GMM Covariance Update
Let d and N denote the dimensionality of your data and the number of observations, respectively.

The following covariance update rule is associated with the Expectation Maximization algorithm for Gaussian Mixture Models:
$$
\Sigma_j = \frac{1}{N_j}
\sum_{n=1}^N \gamma(z_{nj})
(x_n - u_j)(x_n - u_j)^T
$$

where $N_j = \sum_{n=1}^N \gamma(z_{nj})$

Let $\gamma(z_{11}) = 0.3$ and $\gamma(z_{21}) = 0.7$

Compute $\Sigma_1$ and then supply the value of $[\Sigma_1]_{12}$ . (The notation $[X]_{ij}$ refers to the ij-th element of X. In other words i is the row index and j is the column index.)

```python
X = np.array([[8.1,4.2],[9.4,7.7]])
u = np.array([1,1])
gamma = np.array([0.3, 0.7])

def gmm_covariance_update(X, mean, responsibilities):
    """Compute one GMM covariance update from data, mean, and responsibilities."""
    X = np.asarray(X, dtype=float)
    mean = np.asarray(mean, dtype=float)
    responsibilities = np.asarray(responsibilities, dtype=float)

    N_j = np.sum(responsibilities)
    sigma = np.zeros((X.shape[0], X.shape[0]), dtype=float)

    for n in range(X.shape[1]):
        diff = X[:, n] - mean
        sigma += responsibilities[n] * np.outer(diff, diff)

    sigma = sigma / N_j
    return sigma


def gmm_covariance_entry(X, mean, responsibilities, i, j):
    """Return the (i, j) entry of one GMM covariance update."""
    sigma = gmm_covariance_update(X, mean, responsibilities)
    return sigma[i, j]



sigma_1 = gmm_covariance_update(X, u, gamma)
print(sigma_1)

entry_12 = gmm_covariance_entry(X, u, gamma, 0, 1)
print(entry_12)
```

# GMM Utils:
---

```python
def gmm_component_weight(responsibilities):
    """Compute N_j = sum_n gamma(z_nj)."""
    responsibilities = np.asarray(responsibilities, dtype=float)
    return np.sum(responsibilities)


def gmm_mixture_coefficient(responsibilities):
    """Compute pi_j from one responsibility vector."""
    responsibilities = np.asarray(responsibilities, dtype=float)
    return np.sum(responsibilities) / len(responsibilities)


def gmm_mean_update(X, responsibilities):
    """Compute one GMM mean update from data and responsibilities."""
    X = np.asarray(X, dtype=float)
    responsibilities = np.asarray(responsibilities, dtype=float)

    N_j = np.sum(responsibilities)
    mean = np.zeros(X.shape[0], dtype=float)

    for n in range(X.shape[1]):
        mean += responsibilities[n] * X[:, n]

    mean = mean / N_j
    return mean


def gmm_covariance_update(X, mean, responsibilities):
    """Compute one GMM covariance update from data, mean, and responsibilities."""
    X = np.asarray(X, dtype=float)
    mean = np.asarray(mean, dtype=float)
    responsibilities = np.asarray(responsibilities, dtype=float)

    N_j = np.sum(responsibilities)
    sigma = np.zeros((X.shape[0], X.shape[0]), dtype=float)

    for n in range(X.shape[1]):
        diff = X[:, n] - mean
        sigma += responsibilities[n] * np.outer(diff, diff)

    sigma = sigma / N_j
    return sigma


def gmm_covariance_entry(X, mean, responsibilities, i, j):
    """Return the (i, j) entry of one GMM covariance update."""
    sigma = gmm_covariance_update(X, mean, responsibilities)
    return sigma[i, j]


def multivariate_gaussian_pdf(x, mean, covariance):
    """Compute N(x | mean, covariance)."""
    x = np.asarray(x, dtype=float)
    mean = np.asarray(mean, dtype=float)
    covariance = np.asarray(covariance, dtype=float)

    d = len(x)
    diff = x - mean
    inv_cov = np.linalg.inv(covariance)
    det_cov = np.linalg.det(covariance)

    normalizer = 1.0 / np.sqrt(((2.0 * np.pi) ** d) * det_cov)
    exponent = -0.5 * (diff.T @ inv_cov @ diff)

    return float(normalizer * np.exp(exponent))


def gmm_responsibilities_for_point(x, weights, means, covariances):
    """Compute the full responsibility vector for one observation."""
    x = np.asarray(x, dtype=float)
    weights = np.asarray(weights, dtype=float)
    means = np.asarray(means, dtype=float)
    covariances = np.asarray(covariances, dtype=float)

    numerators = np.zeros(len(weights), dtype=float)

    for j in range(len(weights)):
        numerators[j] = weights[j] * multivariate_gaussian_pdf(
            x,
            means[:, j],
            covariances[j]
        )

    denominator = np.sum(numerators)
    return numerators / denominator


def gmm_responsibility_for_component(x, weights, means, covariances, j):
    """Compute gamma(z_j) for one observation and one component."""
    gamma = gmm_responsibilities_for_point(x, weights, means, covariances)
    return gamma[j]


def gmm_log_likelihood(X, weights, means, covariances):
    """Compute the GMM log-likelihood for a full dataset."""
    X = np.asarray(X, dtype=float)
    weights = np.asarray(weights, dtype=float)
    means = np.asarray(means, dtype=float)
    covariances = np.asarray(covariances, dtype=float)

    log_likelihood = 0.0

    for n in range(X.shape[1]):
        mixture_value = 0.0
        for j in range(len(weights)):
            mixture_value += weights[j] * multivariate_gaussian_pdf(
                X[:, n],
                means[:, j],
                covariances[j]
            )
        log_likelihood += np.log(mixture_value)

    return float(log_likelihood)


def gmm_e_step(X, weights, means, covariances):
    """Compute the full responsibility matrix for a dataset."""
    X = np.asarray(X, dtype=float)
    responsibilities = []

    for n in range(X.shape[1]):
        gamma_n = gmm_responsibilities_for_point(X[:, n], weights, means, covariances)
        responsibilities.append(gamma_n)

    responsibilities = np.array(responsibilities).T
    return responsibilities
```
